```python
import numpy as np
from numpy import sin, cos, pi
import matplotlib.pyplot as plt
%matplotlib qt

# -----------------------------------------------------
# System parameters
# -----------------------------------------------------
g = 9.81
m = 1
k = -1
l = 1
F = 1
om = 2/3*pi

# -----------------------------------------------------
# Initial conditions
# -----------------------------------------------------
x0 = 0
v0 = 1

# -----------------------------------------------------
# method parameters
# -----------------------------------------------------
tmax = 300
dt = 0.001
```

```python
# -----------------------------------------------------
# Dynamics: Forced Duffing oscillator
# -----------------------------------------------------
def dyn(t, y):
    x, v = y
    dx = v
    dv = - (k/m)*x - (l/m)*x**3 + (F/m)*cos(om*t)
    return np.array([dx, dv])

# -----------------------------------------------------
# Fourth-order Runge-Kutta method
# -----------------------------------------------------
def rk4(f, t, y, h):
    k1 = h * f(t, y)
    k2 = h * f(t + h/2, y + k1/2)
    k3 = h * f(t + h/2, y + k2/2)
    k4 = h * f(t + h, y + k3)
    return y + (k1 + 2*k2 + 2*k3 + k4) / 6
```

```python
# -----------------------------------------------------
# Integration using RK4
# -----------------------------------------------------
n = int(tmax / dt)
t = np.linspace(0, n * dt, n + 1)
y = np.empty((n + 1, 2))
y[0] = [x0, v0]
for i in range(n):
    y[i + 1] = rk4(dyn, t[i], y[i], dt)

# -----------------------------------------------------
# Separate variables after integration
# -----------------------------------------------------
x = y[:, 0]
v = y[:, 1]

# -----------------------------------------------------
# Phase space
# -----------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, v, linewidth=0.2)
ax.set_xlabel(r'$x$', fontsize=22)
ax.set_ylabel(r'$\dot{x}$', fontsize=22)
ax.tick_params(axis='both', labelsize=20)
#ax.grid(alpha=0.3)
plt.tight_layout()
ax.set_box_aspect(0.65)
plt.savefig("Dufpy.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
